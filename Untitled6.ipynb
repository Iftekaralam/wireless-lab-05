```python

import numpy as np
import matplotlib.pyplot as plt

N = 100000
snr_values = [0, 5, 10, 15, 20]

bits = np.random.randint(0, 2, N)
x = 2 * bits - 1

h = (np.random.randn(N) + 1j * np.random.randn(N)) / np.sqrt(2)

snr_db = 10
snr = 10 ** (snr_db / 10)
noise_var = 1 / snr

n = np.sqrt(noise_var / 2) * (
    np.random.randn(N) + 1j * np.random.randn(N)
)

y = h * x + n
y_eq = y / h

bits_hat = (y_eq.real > 0).astype(int)
ber = np.mean(bits_hat != bits)

print("BER at 10 dB =", ber)

plt.figure(figsize=(8, 5))
plt.hist(np.abs(h), bins=80, density=True)
plt.xlabel("Fading Envelope |h|")
plt.ylabel("Probability Density")
plt.title("Rayleigh Fading Envelope")
plt.grid(True)
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(bits[:50], 'bo-', label="Transmitted Bits")
plt.plot(bits_hat[:50], 'r.--', label="Received Bits")
plt.xlabel("Sample Number")
plt.ylabel("Bit Value")
plt.title("Transmitted and Received Bits")
plt.legend()
plt.grid(True)
plt.show()

ber_values = []

for snr_db in snr_values:
    snr = 10 ** (snr_db / 10)
    noise_var = 1 / snr

    n = np.sqrt(noise_var / 2) * (
        np.random.randn(N) + 1j * np.random.randn(N)
    )

    y = h * x + n
    y_eq = y / h

    bits_hat = (y_eq.real > 0).astype(int)
    ber = np.mean(bits_hat != bits)

    ber_values.append(ber)

    print("SNR =", snr_db, "dB, BER =", ber)

plt.figure(figsize=(8, 5))
plt.semilogy(snr_values, ber_values, 'ro-', linewidth=2, markersize=8)
plt.xlabel("SNR (dB)")
plt.ylabel("Bit Error Rate (BER)")
plt.title("BER vs SNR for Rayleigh Fading")
plt.grid(True)
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

N = 100000
snr_values = [0, 5, 10, 15, 20]

bits = np.random.randint(0, 2, N)
x = 2 * bits - 1

h = (np.random.randn(N) + 1j * np.random.randn(N)) / np.sqrt(2)

ber_awgn = []
ber_rayleigh = []

for snr_db in snr_values:
    snr = 10 ** (snr_db / 10)
    noise_var = 1 / snr

    noise = np.sqrt(noise_var / 2) * (
        np.random.randn(N) + 1j * np.random.randn(N)
    )

    y_awgn = x + noise
    bits_awgn = (y_awgn.real > 0).astype(int)
    ber1 = np.mean(bits_awgn != bits)

    y_rayleigh = h * x + noise
    y_equalized = y_rayleigh / h

    bits_rayleigh = (y_equalized.real > 0).astype(int)
    ber2 = np.mean(bits_rayleigh != bits)

    ber_awgn.append(ber1)
    ber_rayleigh.append(ber2)

    print("SNR =", snr_db, "dB")
    print("AWGN BER =", ber1)
    print("Rayleigh BER =", ber2)
    print("--------------------------------")

plt.figure(figsize=(8, 5))
plt.semilogy(snr_values, ber_awgn, 'bo-', linewidth=2, markersize=8, label="AWGN Only")
plt.semilogy(snr_values, ber_rayleigh, 'ro-', linewidth=2, markersize=8, label="Rayleigh Fading")
plt.xlabel("SNR (dB)")
plt.ylabel("Bit Error Rate (BER)")
plt.title("BER Comparison: AWGN vs Rayleigh Fading")
plt.grid(True)
plt.legend()
plt.show()
```
